```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
baseline = pd.read_csv('/Volumes/2025/2026_paper/COMPARING_SCHEMES/BASELINE/analysis_with_wq_values.csv')
plt.errorbar(baseline['dphi'], baseline['m'], yerr=baseline['dm'], lw=0, ms=6, marker='o', capsize=10, label='baseline', color='black')

colors = sns.color_palette('Blues_d', 4)

for dmu, color in zip([1.0, 1.5, 2.0, 4.0], colors):
    data = pd.read_csv(f'/Volumes/2025/2026_paper/COMPARING_SCHEMES/SCHEME1/D{int(dmu*10)}/analysis_with_wq_values.csv')
    plt.errorbar(data['dphi'], data['m'], yerr=data['dm'], lw=0, ms=6, marker='o', capsize=10, label=f'dmu={dmu}', color=color)

plt.xlabel(r'$\Delta \phi$', fontsize=16)
plt.ylabel(r'$\vert m \vert$', fontsize=16)
plt.title('Homogeneous Driving')
plt.legend()
```
