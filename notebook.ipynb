```python
import numpy as np
import matplotlib.pyplot as plt

results = {'dmu': [0,1,2,4,5], 'dphic': np.array([0.3, 0.564, 0.788, 1.166, 1.347])}
results['dphic'] = results['dphic'] - np.min(results['dphic'])  # Shift to start at 0

dmu_cont = np.linspace(0.1, 6, 1000)
dphi_c = np.log((1+np.exp(1)*np.exp(dmu_cont))/(1+np.exp(dmu_cont)*np.exp(np.exp(-4))))
# dphi_c = dphi_c - np.min(dphi_c)  # Shift to start at 0

# plt.plot(results['dmu'], results['dphic'], 'o', label='Data')
plt.plot(dmu_cont, dphi_c, label='Fit: $\\phi_c = \\frac{1+e\\cdot d\\mu}{1+d\\mu e^{-4}}$')
plt.xlabel('$\\Delta \\mu$')
plt.ylabel('$\\Delta \\Delta \\phi_c$')
```

```python
results = {'dmu': [0,1,2,4,5,6,8], 'dphic': np.array([0.3, 0.564, 0.788, 1.166, 1.347,1.686, 2.336])}
results['dphic'] = results['dphic'] - np.min(results['dphic'])  # Shift to start at 0
plt.plot(results['dmu'], results['dphic'], 'o', label='Data')


from scipy.stats import linregress

res = linregress(results['dmu'], results['dphic'])
print(f"Slope: {res.slope}, Intercept: {res.intercept}, R-squared: {res.rvalue**2}")
plt.plot(results['dmu'], res.intercept + res.slope * np.array(results['dmu']), label='Linear Fit')

plt.xlabel('$\\Delta \\mu$')
plt.ylabel('$\\Delta \\Delta \\phi_c$')
```
